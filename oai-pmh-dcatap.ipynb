```python
import requests
import xml.etree.ElementTree as ET
import rdflib

rdf_ns = "http://www.w3.org/1999/02/22-rdf-syntax-ns#"

response = requests.get('https://portal.odissei.nl/oai?verb=GetRecord&identifier=doi:10.17026/DANS-222-GVUS&metadataPrefix=dcat3-rdfxml')

root = ET.fromstring(response.text)
rdf_element = root.find(f".//{{{rdf_ns}}}RDF")

# print(ET.tostring(rdf_element, encoding='unicode', method='xml'))
rdf_graph = rdflib.Graph()
rdf_graph.parse(data=ET.tostring(rdf_element, encoding='unicode', method='xml'), format='xml')
print(rdf_graph.serialize(format='turtle'))
```
