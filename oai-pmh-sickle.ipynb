```python
# available formats
from sickle import Sickle
sickle = Sickle('https://portal.odissei.nl/oai')
formats = sickle.ListMetadataFormats()
formats_list = [format for format in formats]
for i in formats_list:
    print(dict(i))


# https://portal.odissei.nl/api/datasets/export?exporter=dcat3-rdfxml&persistentId=doi%3A10.17026/SS/5IEOPC
```

```python
from pprint import pprint

records = sickle.ListRecords(metadataPrefix='oai_dc')
for i, rec in enumerate(records):
    if i > 2:
        break
    print(rec.header)
    pprint(rec.metadata)
    
```

```python
# NOTE: dataverse_json is empty
# as by default, Sickle’s mapping of the record XML into Python dictionaries is tailored to work only with Dublin-Core-encoded metadata payloads. 
# Other formats most probably won’t be mapped correctly, 

dataverse_json = sickle.GetRecord(identifier='doi:10.17026/DANS-222-GVUS', metadataPrefix='dataverse_json')

print(f'{"*"*20} dataverse_json {"*"*20}')
pprint(dict(dataverse_json))
```
